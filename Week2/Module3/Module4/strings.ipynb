```python
# 1. Input email text to analyze
email_text = "URGENT! Click here to claim your free gift now!"

# 2. Normalize text to lowercase to ensure case-insensitive matching
lower_text = email_text.lower()

# 3. Use boolean math (True = 1, False = 0) to evaluate spam triggers
spam_score = (
    ("free gift" in lower_text) +
    ("click here" in lower_text) +
    ("buy now" in lower_text) +
    ("earn money" in lower_text) +
    ("act now" in lower_text)
)

# 4. Determine spam status as a Boolean flag (True if score > 0, otherwise False)
is_spam = spam_score > 0

# 5. Map the Boolean result directly to the output text using a dictionary
result_messages = {
    True: f"Flagged as SPAM (Trigger count: {spam_score})",
    False: "Clean Email: No common spam trigger words detected."
}

# 6. Print the mapped message
print(result_messages[is_spam])
```
